```python
import kagglehub

path = kagglehub.dataset_download("dylanjcastillo/7k-books-with-metadata")

print("Path to dataset files:", path)
```

```python
import pandas as pd

books = pd.read_csv(f"{path}/books.csv")
books
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

ax = plt.axes()
sns.heatmap(books.isna().transpose(), cbar=False, ax=ax)

plt.xlabel("Columns")
plt.ylabel("Missing values")
plt.show()
```

```python
import numpy as np
books["missing_description"] = np.where(books["description"].isna(),1,0)
books["age_of_book"] = 2025 - books["published_year"]
```

```python
column_of_interest = ["num_pages" , "age_of_book" , "missing_description" , "average_rating"]
correlation_matrix = books[column_of_interest].corr(method = "spearman")

sns.set_theme(style="white")
plt.figure(figsize=(8,6))
heatmap = sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm",
    cbar_kws={"label": "spearman Correlation"})
heatmap.set_title("correlation heatmap")
plt.show()
```

```python
missing_books = books [~(books["description"].isna()) &
     ~ (books["num_pages"].isna()) &
     ~ (books["average_rating"].isna()) &
     ~ (books["published_year"].isna()) ]
missing_books
```

```python
missing_books["categories"].value_counts().reset_index().sort_values( "count" , ascending = False)
```

```python
missing_books["words_in_description"] = missing_books["description"].str.split().str.len()
missing_books
```

```python
missing_books_25_words = missing_books[missing_books["words_in_description"] > 25]
missing_books_25_words
```

```python
missing_books_25_words["title_and_subtitle"] = (
    np.where(missing_books_25_words["subtitle"].isna(), missing_books_25_words["title"],
    missing_books_25_words[["title", "subtitle"]].astype(str).aggregate(" - ".join, axis=1)))
missing_books_25_words
```

```python
missing_books_25_words["tagged_description"] = missing_books_25_words[["isbn13", "description"]].astype(str).aggregate(" ".join, axis=1)
missing_books_25_words
```

```python
(
    missing_books_25_words
    .drop(["subtitle", "missing_description", "age_of_book", "words_in_description"], axis=1 )
    .to_csv("books_cleaned.csv", index=False, encoding="utf-8")
)
```
